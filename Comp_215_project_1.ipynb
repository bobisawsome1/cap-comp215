```python
import random
```

We start off by creating a dictionay for our system to refer to. We statr by taking the first word and loop through each word after it and adde the words after those to a list keyed by the other 2 words. This means when we input words later on we have more veriety to pick from

```python
#Createing a dictionary
def create_dictionary(filename):
  '''Take a file of text and create a diction ary that will be used topredict the next word is a sequence'''

  #Open the file and create a list with each word
  input_text = open(filename, 'r').readline()
  text_words = input_text.split()

  dictionary = {}

  #Loop through our text and
  for i in range(0, len(text_words) - 1):
    for j in range(i + 1, len(text_words) - 1):
      dictionary[(text_words[i], text_words[j])] = text_words[j + 1:]

  #return our dictionary with all the pairs of prediction
  return dictionary

create_dictionary('test.txt')
```

Next up in a function to predict the next word in a sequence.
We do this by taking 2 words and a dictionary created above and searching for the touple key of the inputted words. Once we have gotten oour list of random words, we randomly pick one from that list.

```python
#Predicting next words
def get_next_word(word_1, word_2, word_dictionary):
  '''Take an input of 2 words and a dictionary of next word predictions and returns a word that comes after them'''

  #Create the touple we will search our dictionary with
  search_touple = (word_1, word_2)

  #Get our list of word to pick from for out prediction
  word_bank = word_dictionary[search_touple]

  #Randomly return on of the words that comes after our inputted words
  return word_bank[random.randint(0, len(word_bank)-1)]

test_dict = create_dictionary('test.txt')
print(get_next_word('I', 'hope', test_dict))
print(get_next_word('I', 'hope', test_dict))
print(get_next_word('I', 'hope', test_dict))
print(get_next_word('This', 'is', test_dict))
```

Using the above function we then start generating sequences of words. We take the last 2 words in the sequence and imput them into our above function and add the output to the sequence. Then repeat with this new word.

```python
#Generate a sequence of predicted words
def create_sequence(start_word_1, start_word_2, text_dictionary, predict_steps):
  '''Takes a pair of starting words and returns the string with the predictions for the next steps'''

  #Get the last word in out sequence so we know when to stop searching
  last_word = text_dictionary[(start_word_1, start_word_2)][-1]
  print(last_word)

  #Place our starting values into our output
  sequence = f'{start_word_1} {start_word_2}'
  for i in range(predict_steps):
    #Take the last 2 words and search for and place our next word in the string
    search_word_1 = sequence.split()[-2]
    print(search_word_1)
    search_word_2 = sequence.split()[-1]
    print(search_word_2)
    if search_word_2 != last_word:
      sequence = sequence + ' ' + get_next_word(search_word_1, search_word_2, text_dictionary)

  return sequence

seq_dic = create_dictionary('test.txt')
create_sequence('This', 'is', seq_dic, 10)
```
